```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, TextStreamer
```

```python
model_id = "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
model = AutoModelForCausalLM.from_pretrained(
    model_id, 
    torch_dtype=torch.bfloat16, 
    device_map="auto"
)
```

```python
streamer = TextStreamer(tokenizer, skip_prompt=True)
```

```python
messages = [{"role": "user", "content": "How to fine tune a LLM? Explain your counting logic step-by-step."}]
inputs = tokenizer.apply_chat_template(
    messages, 
    tokenize=True, 
    add_generation_prompt=True, 
    return_tensors="pt"
).to(model.device)
```

```python
print("\n--- Model Output (Including <think> process) ---\n")
_ = model.generate(
    inputs, 
    streamer=streamer, 
    max_new_tokens=1024,
    temperature=0.6 # DeepSeek-R1 models perform best with temperatures between 0.5 and 0.7
)
```
