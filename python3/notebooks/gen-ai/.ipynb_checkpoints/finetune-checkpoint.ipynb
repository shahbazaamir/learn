```python
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python
from transformers import BitsAndBytesConfig
```

```python
import torch
```

```python
from peft import LoraConfig,  get_peft_model , prepare_model_for_kbit_training
```

```python
model_path='deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B'
```

```python
bnb_config=BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type='nf4',
    bnb_4bit_compute_dtype=torch.bfloat16)
```

```python
model = AutoModelForCausalLM.from_pretrained(model_path)
tokenizer = AutoTokenizer.from_pretrained(model_path)
```

```python
# 2. Prepare the input text
prompt = "Explain quantum computing in one sentence."

# 3. Convert text to numerical token IDs
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

# 4. Generate the response
with torch.no_grad():
    outputs = model.generate(
        **inputs, 
        max_new_tokens=100,  # Control maximum length of output tokens
        do_sample=True,      # Enable creative sampling instead of greedy decoding
        temperature=0.7      # Control randomness (lower is more deterministic)
    )

# 5. Extract and decode the generated text back to English
# Skip special tokens removes technical indicators like <bos> or <eos>
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(generated_text)
```
