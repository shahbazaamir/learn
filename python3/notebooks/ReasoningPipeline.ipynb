```python
import torch
from transformers import pipeline

```

```python

# Load the model with automatic hardware acceleration
reasoning_pipe = pipeline(
    "text-generation", 
    model="deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B", 
    torch_dtype=torch.bfloat16, 
    device_map="auto"
)
```

```python

# Ask a question that requires deep logical reasoning
prompt = "If a tree falls in a forest and no one is around to hear it, does it make a sound? Answer using physical laws vs philosophical laws."

messages = [{"role": "user", "content": prompt}]
outputs = reasoning_pipe(messages, max_new_tokens=512)

print(outputs[0]["generated_text"][-1]["content"])
```

```python
import torch
from transformers import pipeline

# Load the model with automatic hardware acceleration
reasoning_pipe = pipeline(
    "text-generation", 
    model="deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B", 
    torch_dtype=torch.bfloat16, 
    device_map="auto"
)

# Ask a question that requires deep logical reasoning
prompt = "If a tree falls in a forest and no one is around to hear it, does it make a sound? Answer using physical laws vs philosophical laws."

messages = [{"role": "user", "content": prompt}]
outputs = reasoning_pipe(messages, max_new_tokens=512)

print(outputs[0]["generated_text"][-1]["content"])
```
