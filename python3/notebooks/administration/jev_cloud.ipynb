````python
import requests
import os

API_KEY = os.environ["JEV_API_KEY"]

url = "https://api.typesafe.ai/v1/systemone"

payload = {
  "model": "jev-latest",
  "state": "Hi, I have been trying to connect my Stripe account for 3 days and the integration keeps failing. Please help ASAP.",
  "questions": {
    "department": {
      "type": "choice",
      "instructions": "Route this ticket to the correct department.",
      "criteria": {
        "billing": "Issues with payments, invoices, or Stripe setup problems.",
        "integrations": "Technical API errors, webhook issues, or broken connections.",
        "support": "General customer service inquiries, usage questions, or basic assistance."
      }
    },
    "is_urgent": {
      "type": "noul",
      "instructions": "The message conveys high urgency or time sensitivity."
    },
    "frustration_score": {
      "type": "score",
      "instructions": "Rate how upset the customer sounds.",
      "criteria": [
        "Calm, just stating facts.",
        "Frustrated but polite.",
        "Very angry, using intense language or all caps."
      ]
    }
  }
}

headers = {
    "Authorization": f"Bearer {API_KEY}",
    "Content-Type": "application/json"
}

response = requests.post(url, json=payload, headers=headers)

print(response.json())
````
