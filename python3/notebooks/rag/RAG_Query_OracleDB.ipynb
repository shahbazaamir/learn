```python
import os
import oracledb

connection = oracledb.connect(
    user=os.environ["ORACLE_USER"],
    password=os.environ["ORACLE_PASS"],
    dsn=os.environ["ORACLE_TLS_DNS"]
)
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_oracledb.vectorstores.oraclevs import OracleVS
from langchain_oracledb.vectorstores import DistanceStrategy

embedding_model = OllamaEmbeddings(
    model="nomic-embed-text"
)

vector_store = OracleVS(
    client=connection,
    table_name="DOCUMENT_VECTORS",
    embedding_function=embedding_model,
    distance_strategy=DistanceStrategy.COSINE
)

```

```python
results = vector_store.similarity_search(
    "What is Oracle?",
    k=1
)

for r in results:
    print(r.metadata)
    print(r.page_content)
```

```python
vector_store2 = OracleVS(
    client=connection,
    table_name="SOLUTION_ARCHITECT_VECTORS",
    embedding_function=embedding_model,
    distance_strategy=DistanceStrategy.COSINE
)
```

```python
query2 = "What is DRM?"

results2 = vector_store2.similarity_search(
    query2,
    k=2
)

for i, result in enumerate(results2, 1):
    print(f"\nResult {i}")
    print("Content:", result.page_content)
    print("Metadata:", result.metadata)
```
